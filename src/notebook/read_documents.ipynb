```python
import pandas as pd
import numpy as np


#variables = variables_eda + ["COMPRA"]

#columnas = [c for c in variables if c in df.columns]

#df_eda = df[columnas].copy()

df_eda = pd.read_csv("data/bronce/df_bronce.csv")

# Convertir fechas
fechas = [
    "FECHA_DE_CREACION_OP",
    "FECHA_ULTIMA_ACTUALIZACION_OP",
    "FECHA_INICIO_VENTAS_TRANSACCIONAL"
]

for col in fechas:
    if col in df_eda.columns:
        df_eda[col] = pd.to_datetime(
            df_eda[col],
            errors="coerce"
        )

# Fecha de referencia para análisis retrospectivo
fecha_corte = df_eda["FECHA_DE_CREACION_OP"].max()

# Crear variables temporales
if pd.notna(fecha_corte):
    if "FECHA_DE_CREACION_OP" in df_eda:
        df_eda["ANTIGUEDAD_OP_DIAS"] = (
            fecha_corte - df_eda["FECHA_DE_CREACION_OP"]
        ).dt.days

    if "FECHA_ULTIMA_ACTUALIZACION_OP" in df_eda:
        df_eda["DIAS_ULTIMA_ACTUALIZACION"] = (
            fecha_corte - df_eda["FECHA_ULTIMA_ACTUALIZACION_OP"]
        ).dt.days

    if "FECHA_INICIO_VENTAS_TRANSACCIONAL" in df_eda:
        df_eda["ANTIGUEDAD_PROYECTO_DIAS"] = (
            fecha_corte - df_eda["FECHA_INICIO_VENTAS_TRANSACCIONAL"]
        ).dt.days
```

```python
from scipy.stats import (
    chi2_contingency,
    pointbiserialr
)

def cramers_v(x, y):
    tabla = pd.crosstab(x, y)

    if min(tabla.shape) < 2:
        return np.nan

    chi2 = chi2_contingency(
        tabla,
        correction=False
    )[0]

    n = tabla.to_numpy().sum()
    phi2 = chi2 / n

    r, k = tabla.shape

    phi2_corr = max(
        0,
        phi2 - ((k - 1) * (r - 1)) / (n - 1)
    )

    r_corr = r - ((r - 1)**2) / (n - 1)
    k_corr = k - ((k - 1)**2) / (n - 1)

    denominador = min(
        k_corr - 1,
        r_corr - 1
    )

    if denominador <= 0:
        return np.nan

    return np.sqrt(phi2_corr / denominador)
```

```python
variables_categoricas = [
    "ETAPA_DE_VENTA",
    "REFERIDO",
    "TIPO_DE_PROYECTO",
    "METODO_DE_VENTAS",
    "CIUDAD",
    "DESTINO",
    "TIPO_VENTA",
    "SALA_DE_VENTAS_ALTERNATIVA",
    "ESTADO_PROYECTO",
    "BK_VENDEDOR",
    "OCUPACION_COMPRADOR1",
    "CIUDAD_COMPRADOR1"
]

variables_numericas = [
    "IMPORTE_TOTAL_OP_PRIMERA_VERSION",
    "ANTIGUEDAD_OP_DIAS",
    "DIAS_ULTIMA_ACTUALIZACION",
    "ANTIGUEDAD_PROYECTO_DIAS"
]

resultados = []

for col in variables_categoricas:

    if col not in df_eda.columns:
        continue

    datos = df_eda[[col, "COMPRA"]].dropna()

    if datos.empty:
        continue

    asociacion = cramers_v(
        datos[col],
        datos["COMPRA"]
    )

    resultados.append({
        "Variable": col,
        "Metodo": "V de Cramér",
        "Asociacion": asociacion,
        "N": len(datos)
    })


for col in variables_numericas:

    if col not in df_eda.columns:
        continue

    datos = df_eda[[col, "COMPRA"]].copy()

    datos[col] = pd.to_numeric(
        datos[col],
        errors="coerce"
    )

    datos = datos.replace(
        [np.inf, -np.inf],
        np.nan
    ).dropna()

    if (
        len(datos) < 3
        or datos[col].nunique() < 2
        or datos["COMPRA"].nunique() < 2
    ):
        continue

    correlacion, p_value = pointbiserialr(
        datos["COMPRA"],
        datos[col]
    )

    resultados.append({
        "Variable": col,
        "Metodo": "Punto-biserial",
        "Asociacion": correlacion,
        "N": len(datos)
    })

df_correlaciones = pd.DataFrame(resultados)

df_correlaciones["Magnitud"] = (
    df_correlaciones["Asociacion"].abs()
)

df_correlaciones = df_correlaciones.sort_values(
    "Magnitud",
    ascending=False
)

display(df_correlaciones)
```
